```python
# Define all agents with valid names 

# Customer Proxy Agent
customer_proxy_agent = autogen.ConversableAgent(
    name="Customer_Proxy", 
    system_message="Acts as an intermediary to relay customer requirements to the Project Manager.",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="ALWAYS",
)

# Project Manager Agent
project_manager_agent = autogen.ConversableAgent(
    name="Project_Manager",  
    system_message="Receives customer requirements and formulates a project plan by integrating inputs from various engineers.",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="ALWAYS",
)

# Requirement Engineer Agent
requirement_engineer_agent = autogen.ConversableAgent(
    name="Requirement_Engineer",  
    system_message= """ First, drafts a requirements document based on customer needs for the fitness tracker and diet planning system. 
                        
                        Collaborates with the Fitness Agent, Dietetic Agent, and Data Acquisition Agent to collect specific requirements:
                            - The Fitness Agent identifies fitness objectives, preferred workout styles, and tracking methods.
                            - The Dietetic Agent gathers nutritional preferences, daily calorie goals, and meal planning needs.
                            - The Data Acquisition Agent collects key user data, including age, weight, and activity level, to enable personalized recommendations.
                        
                        Then, calculates the {effort} required to complete the requirements document in hours, days, weeks, or months by following these steps:
                            - Step 1: work = {estimate total number of fitness and diet-related requirements collected} 
                            - Step 2: productivity rate = {5 requirements documented per day} 
                            - Step 3: effort = {work}/{productivity} .
                        
                        The total {work} in this phase is determined by the number of gathered and documented fitness and dietary requirements. """,

    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="ALWAYS",
)

# Fitness Agent
fitness_agent = autogen.ConversableAgent(
    name="Fitness_Agent",  
    system_message="Collects user fitness goals, including workout types (cardio, strength), exercise frequency, and progress tracking preferences.",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="ALWAYS",
)

# Dietetic Agent
dietetic_agent = autogen.ConversableAgent(
    name="Dietetic_Agent",  
    system_message="Collects user dietary requirements, such as calorie targets, meal preferences (e.g., vegetarian, keto), and meal planning needs.",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="ALWAYS",
)

# Data Acquisition Agent
data_acquisition_agent = autogen.ConversableAgent(
    name="Data_Acquisition_Agent",  
    system_message="Gathers essential user data, including age, weight, height, and activity level, to enable personalized fitness and dietary recommendations.",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="ALWAYS",
)

# System Engineer Agent
system_engineer_agent = autogen.ConversableAgent(
    name="System_Engineer",  
    system_message="Develops a detailed design document based on the requirements document provided by the Requirement Engineer. Computes the {effort} required for design creation.",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="ALWAYS",
)

# Software Engineer Agent
software_engineer_agent = autogen.ConversableAgent(
    name="Software_Engineer",  
    system_message="Implements the complete software system based on the design document, estimating the workload and effort using source lines of code (SLOC).",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="ALWAYS",
)

# Test Engineer Agent
test_engineer_agent = autogen.ConversableAgent(
    name="Test_Engineer",  
    system_message="Designs and executes test plans, calculating the {effort} required to develop and run test cases.",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="ALWAYS",
)

# Documentation Engineer Agent
documentation_engineer_agent = autogen.ConversableAgent(
    name="Documentation_Engineer",  
    system_message="Prepares comprehensive user documentation and training materials based on the software and design specifications. Estimates the {effort} for this phase.",
    code_execution_config=False,
    llm_config=llm_config,
    human_input_mode="ALWAYS",
)
```

```python
# Define the nested conversation flow
nested_chat_queue = [
    {"recipient": requirement_engineer_agent, "message": "Document comprehensive use-cases and specifications for the web-based fitness monitoring app."},
    {"recipient": system_engineer_agent, "message": "Develop the system architecture based on the finalized requirements document."},
    {"recipient": software_engineer_agent, "message": "Implement the software according to the system design specifications."},
    {"recipient": test_engineer_agent, "message": "Design and execute test cases to validate functionality and performance."},
    {"recipient": documentation_engineer_agent, "message": "Prepare user guides and documentation based on the developed system."},
]
```

```python
# To store the output from each agent
engineer_outputs = {}

# Function to extract the work of each engineer from the chat result
def extract_engineer_work(chat_result):
    # Extract only the response from the agent (output)
    if chat_result and chat_result.chat_history:
        return chat_result.chat_history[-1]['content']
    return 'No output available'
```

```python
def initiate_review_and_feedback_chats():
    accumulated_work = ""
    engineer_outputs = {}

    # Go through the nested chat queue and insert a review cycle
    for chat in nested_chat_queue:
        recipient = chat['recipient']
        message = chat['message']

        # Include accumulated work in the message for the next engineer
        if accumulated_work:
            message += f"\n\nHere is the combined work from previous phases:\n{accumulated_work}"

        print(f"Project Manager initiates chat with {recipient.name.replace('_', ' ')}")

        # Initiate chat and capture the response
        output = project_manager_agent.initiate_chat(
            recipient=recipient,
            message=message
        )

        # Extract the agent's work from the response
        response = extract_engineer_work(output)

        # Now, instead of review_engineer_work, the Project Manager reviews the work
        review = project_manager_agent.review_work(response)
        if review.needs_revision:
            print(f"Requesting revision from {recipient.name}")
            continue  # Send it back for revision if needed

        # Store the response from each agent
        engineer_outputs[recipient.name] = {
            "speaker": project_manager_agent.name,
            "recipient": recipient.name,
            "input_message": message,
            "response": response
        }

        # Append the current engineer's work to the accumulated work
        accumulated_work += f"\n\nWork from {recipient.name.replace('_', ' ')}:\n{response}"

    # Compile the final project plan after review and feedback cycles
    final_message_to_customer = compile_final_project_plan(engineer_outputs)

    # Project Manager sends final message to Customer Proxy
    print(f"Project Manager sends the final project plan to Customer Proxy")
    project_manager_agent.initiate_chat(
        recipient=customer_proxy_agent,
        message=final_message_to_customer
    )
```

```python
class ConversableAgent:
    def _init_(self, name):
        self.name = name

    def initiate_chat(self, recipient, message):
        # Simulate a conversation between agents
        print(f"{self.name} is sending a message to {recipient.name}: {message}")
        # Simulate the recipient responding
        return recipient.respond(message)

    def respond(self, message):
        # Simulate a response from the agent
        print(f"{self.name} is responding to the message.")
        return f"{self.name}'s work completed."

    # Add the ⁠ review_work ⁠ method to handle reviewing the work
    def review_work(self, response):
        # Simple logic to simulate review
        print(f"{self.name} is reviewing the work.")
        if "needs_revision" in response:
            return ReviewResult(needs_revision=True)
        else:
            return ReviewResult(needs_revision=False)

class ReviewResult:
    def _init_(self, needs_revision):
        self.needs_revision = needs_revision
```

```python
# Function to handle nested chat flow and accumulate work from each agent
def initiate_nested_chats():
    # Customer Proxy initiates conversation with the Project Manager
    customer_proxy_agent.initiate_chat(
        recipient=project_manager_agent,
        message=customer_message
    )
    
    # To accumulate all previous engineers' work
    accumulated_work = ""

    # Project Manager passes messages and accumulated work to each engineer
    for i, chat in enumerate(nested_chat_queue):
        recipient = chat['recipient']
        message = chat['message']

        # Include accumulated work in the message for the next engineer
        if accumulated_work:
            message += f"\n\nHere is the combined work from previous phases:\n{accumulated_work}"

        print(f"Project Manager initiates chat with {recipient.name.replace('_', ' ')}")
        
        # Initiate chat and capture the response
        output = project_manager_agent.initiate_chat(
            recipient=recipient,
            message=message
        )

        # Extract the agent's work from the response
        response = extract_engineer_work(output)

        # Store the response from each agent
        engineer_outputs[recipient.name] = {
            "speaker": project_manager_agent.name,
            "recipient": recipient.name,
            "input_message": message,
            "response": response
        }

        # Append the current engineer's work to the accumulated work
        accumulated_work += f"\n\nWork from {recipient.name.replace('_', ' ')}:\n{response}"

    # Once all nested chats are complete, compile the final project plan
    final_message_to_customer = compile_final_project_plan(engineer_outputs)

    # Project Manager sends final message to Customer Proxy
    print(f"Project Manager sends the final project plan to Customer Proxy")
    project_manager_agent.initiate_chat(
        recipient=customer_proxy_agent,
        message=final_message_to_customer
    )
```

```python
# Function to compile the final project plan
def compile_final_project_plan(engineer_outputs):
    final_project_plan = "### Final Project Plan\n\n"
    final_project_plan += "All phases of the project have been completed. Below is the final project plan detailing the work done by each engineer:\n\n"
    
    # Initialize chat log index starting from 1
    chat_log_index = 1
    
    # Add a section for each phase
    for engineer, details in engineer_outputs.items():
        final_project_plan += f"#### Chat Log {chat_log_index}: {engineer.replace('_', ' ')} Phase\n"
        final_project_plan += f"- Speaker: {details['speaker'].replace('_', ' ')}\n"
        final_project_plan += f"- Recipient: {details['recipient'].replace('_', ' ')}\n"
        final_project_plan += f"- Input Message:\n{details['input_message']}\n\n"
        final_project_plan += f"- Response:\n{details['response']}\n\n"
        
        # Increment the chat log index for the next engineer
        chat_log_index += 1
    
    final_project_plan += "### Conclusion:\n"
    final_project_plan += "The project is ready for deployment, and all necessary phases have been completed successfully."
    
    return final_project_plan

# Start the nested conversation
initiate_nested_chats()
```

```python
# Define the nested conversation flow
nested_chat_queue = [
    {"recipient": data_acquisition_agent, "message": "Collect essential user data, including age, weight, height, and activity level for personalized insights.", "priority": 3},
    {"recipient": fitness_agent, "message": "Identify key fitness tracking requirements such as workout types, frequency, and progress tracking.", "priority": 4},
    {"recipient": dietetic_agent, "message": "Gather dietary tracking needs, including meal logging, calorie goals, and dietary preferences.", "priority": 2},
    {"recipient": requirement_engineer_agent, "message": "Compile and document detailed use-cases and specifications for the web-based fitness monitoring app.", "priority": 1},
    {"recipient": system_engineer_agent, "message": "Design the system architecture based on the finalized requirements document.", "priority": 5},
    {"recipient": software_engineer_agent, "message": "Develop the software following the system design specifications.", "priority": 6},
    {"recipient": test_engineer_agent, "message": "Plan and execute test cases to validate functionality and ensure reliability.", "priority": 7},
    {"recipient": documentation_engineer_agent, "message": "Create comprehensive user documentation and guides based on the developed system.", "priority": 8},
]
```

```python
def initiate_priority_based_chats():
    accumulated_work = ""
    engineer_outputs = {}
    
    # Assign priority to each task in the nested chat queue
    task_priority_queue = sorted(nested_chat_queue, key=lambda x: x['priority'], reverse=True)

    # Process tasks based on their priority
    for task in task_priority_queue:
        recipient = task['recipient']
        message = task['message']

        # Include accumulated work in the message for the next engineer
        if accumulated_work:
            message += f"\n\nHere is the combined work from previous phases:\n{accumulated_work}"

        print(f"Project Manager initiates priority-based chat with {recipient.name.replace('_', ' ')}")

        # Initiate chat and capture the response
        output = project_manager_agent.initiate_chat(
            recipient=recipient,
            message=message
        )

        # Extract the agent's work from the response
        response = extract_engineer_work(output)

        # Store the response from each agent
        engineer_outputs[recipient.name] = {
            "speaker": project_manager_agent.name,
            "recipient": recipient.name,
            "input_message": message,
            "response": response
        }

        # Append the current engineer's work to the accumulated work
        accumulated_work += f"\n\nWork from {recipient.name.replace('_', ' ')}:\n{response}"

    # Compile the final project plan after priority-based chats
    final_message_to_customer = compile_final_project_plan(engineer_outputs)

    # Project Manager sends final message to Customer Proxy
    print(f"Project Manager sends the final project plan to Customer Proxy")
    project_manager_agent.initiate_chat(
        recipient=customer_proxy_agent,
        message=final_message_to_customer
    )
```

```python
# import autogen
# import os
# import openai

# # Ensure API key is set
# openai.api_key = os.getenv("OPENAI_API_KEY")

# # Define agents
# customer_proxy_agent = autogen.ConversableAgent(
#     name="Customer_Proxy", 
#     system_message="Acts as an intermediary to relay customer requirements to the Project Manager.",
#     code_execution_config=False,
#     llm_config={"model": "gpt-3.5-turbo"},
#     human_input_mode="ALWAYS",
# )

# project_manager_agent = autogen.ConversableAgent(
#     name="Project_Manager",  
#     system_message="Receives customer requirements and formulates a project plan by integrating inputs from various engineers.",
#     code_execution_config=False,
#     llm_config={"model": "gpt-3.5-turbo"},
#     human_input_mode="ALWAYS",
# )

# # Define engineers (examples)
# requirement_engineer_agent = autogen.ConversableAgent(
#     name="Requirement_Engineer", 
#     system_message="Analyzes and documents software requirements.",
#     code_execution_config=False,
#     llm_config={"model": "gpt-3.5-turbo"},
#     human_input_mode="ALWAYS",
# )

# design_engineer_agent = autogen.ConversableAgent(
#     name="Design_Engineer", 
#     system_message="Transforms requirements into a structured design.",
#     code_execution_config=False,
#     llm_config={"model": "gpt-3.5-turbo"},
#     human_input_mode="ALWAYS",
# )

# # Sample nested chat queue with priorities
# nested_chat_queue = [
#     {"recipient": requirement_engineer_agent, "message": "Define system requirements", "priority": 3},
#     {"recipient": design_engineer_agent, "message": "Create the system design based on requirements", "priority": 2},
# ]

# def extract_engineer_work(output):
#     """Extracts useful work details from the chat output."""
#     return output.get("response", "No response received.")

# def compile_final_project_plan(engineer_outputs):
#     """Compiles the collected work from engineers into a project plan."""
#     project_plan = "Final Project Plan:\n"
#     for engineer, data in engineer_outputs.items():
#         project_plan += f"\nFrom {engineer}:\n{data['response']}\n"
#     return project_plan

# def initiate_priority_based_chats():
#     accumulated_work = ""
#     engineer_outputs = {}
    
#     # Assign priority to each task in the queue
#     task_priority_queue = sorted(nested_chat_queue, key=lambda x: x['priority'], reverse=True)

#     # Process tasks based on priority
#     for task in task_priority_queue:
#         recipient = task['recipient']
#         message = task['message']

#         # Include accumulated work from previous phases
#         if accumulated_work:
#             message += f"\n\nHere is the combined work from previous phases:\n{accumulated_work}"

#         print(f"Project Manager initiates chat with {recipient.name.replace('_', ' ')}")

#         # Initiate chat and capture response
#         output = project_manager_agent.initiate_chat(
#             recipient=recipient,
#             message=message
#         )

#         # Extract engineer's work
#         response = extract_engineer_work(output)
#         engineer_outputs[recipient.name] = {
#             "speaker": project_manager_agent.name,
#             "recipient": recipient.name,
#             "input_message": message,
#             "response": response
#         }

#         # Append current engineer's work to accumulated work
#         accumulated_work += f"\n\nWork from {recipient.name.replace('_', ' ')}:\n{response}"

#     # Compile final project plan
#     final_message_to_customer = compile_final_project_plan(engineer_outputs)
#     print("Project Manager sends final project plan to Customer Proxy")
    
#     # Send final plan to Customer Proxy
#     project_manager_agent.initiate_chat(
#         recipient=customer_proxy_agent,
#         message=final_message_to_customer
#     )

# # Run function
# initiate_priority_based_chats()

import autogen
import os
import openai

# Ensure API key is set
openai.api_key = os.getenv("OPENAI_API_KEY")

# Define agents
customer_proxy_agent = autogen.ConversableAgent(
    name="Customer_Proxy", 
    system_message="Acts as an intermediary to relay customer requirements to the Project Manager.",
    code_execution_config=False,
    llm_config={"model": "gpt-3.5-turbo"},
    human_input_mode="ALWAYS",
)

project_manager_agent = autogen.ConversableAgent(
    name="Project_Manager",  
    system_message="Receives customer requirements and formulates a project plan by integrating inputs from various engineers.",
    code_execution_config=False,
    llm_config={"model": "gpt-3.5-turbo"},
    human_input_mode="ALWAYS",
)

# Define engineers (examples)
requirement_engineer_agent = autogen.ConversableAgent(
    name="Requirement_Engineer", 
    system_message="Analyzes and documents software requirements.",
    code_execution_config=False,
    llm_config={"model": "gpt-3.5-turbo"},
    human_input_mode="ALWAYS",
)

design_engineer_agent = autogen.ConversableAgent(
    name="Design_Engineer", 
    system_message="Transforms requirements into a structured design.",
    code_execution_config=False,
    llm_config={"model": "gpt-3.5-turbo"},
    human_input_mode="ALWAYS",
)

# Sample nested chat queue with priorities
nested_chat_queue = [
    {"recipient": requirement_engineer_agent, "message": "Define system requirements", "priority": 3},
    {"recipient": design_engineer_agent, "message": "Create the system design based on requirements", "priority": 2},
]

def extract_engineer_work(output):
    """Extracts useful work details from the chat output."""
    return getattr(output, "response", "No response received.")

def compile_final_project_plan(engineer_outputs):
    """Compiles the collected work from engineers into a project plan."""
    project_plan = "Final Project Plan:\n"
    for engineer, data in engineer_outputs.items():
        project_plan += f"\nFrom {engineer}:\n{data['response']}\n"
    return project_plan

def initiate_priority_based_chats():
    accumulated_work = ""
    engineer_outputs = {}
    
    # Assign priority to each task in the queue
    task_priority_queue = sorted(nested_chat_queue, key=lambda x: x['priority'], reverse=True)

    # Process tasks based on priority
    for task in task_priority_queue:
        recipient = task['recipient']
        message = task['message']

        # Include accumulated work from previous phases
        if accumulated_work:
            message += f"\n\nHere is the combined work from previous phases:\n{accumulated_work}"

        print(f"Project Manager initiates chat with {recipient.name.replace('_', ' ')}")

        # Initiate chat and capture response
        output = project_manager_agent.initiate_chat(
            recipient=recipient,
            message=message
        )

        # Extract engineer's work
        response = extract_engineer_work(output)
        engineer_outputs[recipient.name] = {
            "speaker": project_manager_agent.name,
            "recipient": recipient.name,
            "input_message": message,
            "response": response
        }

        # Append current engineer's work to accumulated work
        accumulated_work += f"\n\nWork from {recipient.name.replace('_', ' ')}:\n{response}"

    # Compile final project plan
    final_message_to_customer = compile_final_project_plan(engineer_outputs)
    print("Project Manager sends final project plan to Customer Proxy")
    
    # Send final plan to Customer Proxy
    project_manager_agent.initiate_chat(
        recipient=customer_proxy_agent,
        message=final_message_to_customer
    )

# Run function
initiate_priority_based_chats()

```
